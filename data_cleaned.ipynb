```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data.csv")
```

```python
print(df["occasion"].value_counts())
print(df["formality"].value_counts())
print(df["constraint"].value_counts())
print(df["color_tone"].value_counts())
```

```python
df["occasion"].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['occasion_encoded'] = le.fit_transform(df['occasion'])

print(df['occasion_encoded'].value_counts())
```

```python
df["formality"].value_counts()
```

```python
df['formality_encoded'] = le.fit_transform(df['formality'])
print(df['formality_encoded'].value_counts())
```

```python
df['constraint'].value_counts()
```

```python
df['color_tone'].value_counts()
```

```python
df['constraint_encoded'] = le.fit_transform(df['constraint'])
print(df['constraint_encoded'].value_counts())
```

```python
df['color_tone_encoded'] = le.fit_transform(df['color_tone'])
print(df['color_tone_encoded'].value_counts())
```
